## Question 1: Data Types and Input Handling


Take student name, roll number, age, and marks as input.
Display each value with its data type.
Convert age to float and marks to int before displaying.


```python
student_name = input("Enter name of the student")
roll_no = int(input("Enter roll_no of the student"))
age = int(input("Enter the age of the student"))
marks = float(input("Enter marks of the student"))

print(student_name,type(student_name))
print(roll_no,type(roll_no))
print(age,type(age))
print(marks,type(marks))
print("After Type Casting")

age_float = float(age)
marks_int = int(marks)
print(age_float,type(age_float))
print(marks_int,type(marks_int))
```

#Conclusion : used simple type casting to change the datatypes of age and marks to float and int respectively 

## Question 2: Conditional Logic

Accept marks (0–100) and display:

Excellent (≥80)
Good (≥60)
Pass (≥40)
Fail (<40)

```python
marks = float(input("Enter the marks of the student"))
print("marks is :",marks)

if marks >=0 and marks <=100:
    if marks >=80:
        print("Excellent")

    elif marks >=60 and marks <80:
        print("Good")
    elif marks >=40 and marks <60:
        print("pass")
    else:
        print("Fail")
```

Conclusion : Used simple conditional logic using if elseif statement


## Question 3: Looping with Conditions

Print numbers between 1 and 100 that are divisible by 4 but not divisible by 8.

```python
for i in range(1,100):
    if i%4==0 and i%8 !=0:
        print(i)
```

Conclusion : Used Simple loop to meet the required solution
